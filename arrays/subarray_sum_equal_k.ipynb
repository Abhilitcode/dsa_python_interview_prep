

### Example

```text
arr = [3]
K = 3
```

We want to count subarrays whose sum is `3`.

There is one:

```text
[3]
```

So answer should be **1**.

---

### Start

```python
prefix_sum = 0
count = 0
hashmap = {0: 1}
```

The `{0: 1}` means:

> **"Before we start looking at the array, we've already seen a sum of 0 once."**

Why do we pretend we've seen `0`?

Because we are standing **just before index 0**.

Think:

```text
Before array
     ↓
   [3]
    0
```

The sum before taking any elements is naturally **0**.

---

### Now take `3`

```python
prefix_sum += 3
```

So:

```text
prefix_sum = 3
```

We want `K = 3`.

We calculate:

```text
prefix_sum - K
= 3 - 3
= 0
```

Now we ask the hashmap:

> "Have I seen a prefix sum of `0`?"

Yes!

```python
hashmap = {0: 1}
```

It says:

> I've seen `0` **once**.

Therefore:

```python
count += 1
```

So:

```text
count = 1
```

And that `1` represents this subarray:

```text
[3]
```

---

### What if we DON'T put `{0: 1}`?

Suppose we started with:

```python
hashmap = {}
```

Then after seeing `3`:

```text
prefix_sum = 3
```

We calculate:

```text
3 - 3 = 0
```

But:

```text
0 not in hashmap
```

So we **wouldn't count `[3]`**.

That's why `{0: 1}` is there.

### 🧠 Easy way to remember

```python
hashmap = {0: 1}
```

means:

> **"Before the array starts, the sum is 0, and we've seen that starting point once."**

This lets us correctly count subarrays that **start from index 0**.

For example:

```text
[3]
[1, 2]
[1, 2, 3]
```

All of these start at index `0`, and the initial `0` in the hashmap makes the prefix-sum method able to detect them.
