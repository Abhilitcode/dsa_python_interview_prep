# Kadane's Algorithm — What are we actually trying to do?

We have an array:

```text
[-2, 1, -3, 4, -1, 2, 1, -5, 4]
```

We need to find a **continuous/contiguous part** of this array whose sum is the biggest.

For example:

```text
[4, -1, 2, 1]
```

Sum:

```text
4 + (-1) + 2 + 1 = 6
```

So answer = **6**.

---

# First understand what "subarray" means

A subarray means:

> Elements must be **next to each other**.

For:

```text
[1, 2, 3, 4]
```

Valid subarrays include:

```text
[1]
[2]
[3]
[4]

[1,2]
[2,3]
[3,4]

[1,2,3]
[2,3,4]

[1,2,3,4]
```

But this is **not** a subarray:

```text
[1,3]
```

because we skipped `2`.

---

# So what's the problem?

Imagine we are walking through the array from left to right.

Take:

```text
[-2, 1, -3, 4, -1, 2, 1, -5, 4]
```

At every number, we have a decision:

> **Should I continue the subarray I already have, or should I throw it away and start a new subarray from this number?**

THIS is the heart of Kadane's Algorithm.

---

# Let's understand WHY we need that decision

Look at this smaller array:

```text
[-5, 4]
```

Suppose we have already collected:

```text
[-5]
```

Current sum:

```text
-5
```

Now we see `4`.

We have two choices.

### Choice 1 — Continue

```text
-5 + 4 = -1
```

### Choice 2 — Start fresh

Just take:

```text
4
```

Which is better?

```text
4 > -1
```

So obviously we should **throw away -5** and start from `4`.

That's what Kadane does.

---

# Another example

Look at:

```text
[4, -1, 2]
```

We have:

```text
4
```

Then we see:

```text
-1
```

Should we throw away `4`?

Let's check.

### Continue:

```text
4 + (-1) = 3
```

### Start fresh:

```text
-1
```

Obviously:

```text
3 > -1
```

So we **keep 4**.

Our current subarray becomes:

```text
[4, -1]
```

Current sum:

```text
3
```

Then we see `2`.

Continue:

```text
3 + 2 = 5
```

Start fresh:

```text
2
```

Continue is better.

So:

```text
[4, -1, 2]
```

has sum `5`.

---

# 🔥 The simple rule

At every number:

> **Either start from this number OR add this number to the previous subarray.**

Mathematically:

But don't memorize the formula yet.

In simple English:

```text
                    ┌── Start fresh from current number
Current number ─────┤
                    └── Continue previous subarray + current number

             Pick whichever gives bigger sum
```

---

# Now why do we need `max_sum`?

This is a **different thing** from `current_sum`.

Suppose:

```text
[4, -1, 2, 1, -10]
```

We reach:

```text
[4, -1, 2, 1]
```

Current sum:

```text
6
```

So:

```text
max_sum = 6
```

Then we add `-10`:

```text
6 + (-10) = -4
```

Now:

```text
current_sum = -4
```

But did our previous answer of `6` disappear?

**No!**

That's why we keep another variable:

```text
current_sum → best sum ending HERE
max_sum     → best sum found ANYWHERE so far
```

This distinction is extremely important.

---

# Think of them like this

Imagine you're climbing mountains.

### `current_sum`

> "How high am I currently?"

### `max_sum`

> "What is the highest point I have reached during the entire journey?"

You can come down later, but your highest point is still remembered.

---

# So what is Kadane actually doing?

It is **not really "finding the maximum sum" directly**.

It is repeatedly asking:

> **"Is my previous subarray helping me or hurting me?"**

If it helps:

```text
continue
```

If starting fresh is better:

```text
throw away the previous sum
```

And separately:

```text
remember the biggest sum we've ever seen
```

That's Kadane.

---

# Now the code becomes much easier to understand

```python
def max_subarray(arr):

    current_sum = arr[0]
    max_sum = arr[0]

    for i in range(1, len(arr)):

        current_sum = max(arr[i], current_sum + arr[i])

        max_sum = max(max_sum, current_sum)

    return max_sum
```

### The most important line:

```python
current_sum = max(arr[i], current_sum + arr[i])
```

Read it like English:

> **"Should I start a new subarray with `arr[i]`, or continue my existing subarray by adding `arr[i]`? Give me whichever is bigger."**

And:

```python
max_sum = max(max_sum, current_sum)
```

means:

> **"Is my current sum bigger than the best answer I've seen so far? If yes, remember it."**

---

# One final thing you MUST understand

Why don't we just say:

```python
if current_sum < 0:
    current_sum = 0
```

That's a common version of Kadane, but **it can confuse you when all numbers are negative**.

Example:

```text
[-5, -2, -8]
```

The correct answer is:

```text
-2
```

because the subarray `[−2]` has the largest sum.

So for your learning, I recommend this version:

```python
current_sum = max(arr[i], current_sum + arr[i])
```

because it directly expresses the actual decision:

**continue OR start fresh.**

### Remember just these 2 things:

```text
current_sum → best subarray sum ending at the current position

max_sum → best subarray sum found anywhere so far
```
